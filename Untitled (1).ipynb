```python
import pandas as pd
```

```python
movies = pd.read_csv("tmdb_5000_movies.csv")
credits = pd.read_csv("tmdb_5000_credits.csv")
```

```python
data = movies.merge(credits, left_on = "id", right_on = "movie_id")
```

```python
data = data[["title_x", "budget", "genres", "keywords", "spoken_languages", "cast", "crew", "overview"]]
```

```python
import ast
def extract_1(obj):
    
    l = []
    for i in ast.literal_eval(obj):
        l.append(i["name"])
    return l
```

```python
import ast

def extract_2(obj):

    if isinstance(obj, str):
        obj = ast.literal_eval(obj)
        
    d = []
    for i in obj[:7]:
        d.append(i["character"] + ":" + i["name"])

    return d
```

```python
import ast

def extract_3(obj):
    if isinstance(obj, str):
        obj = ast.literal_eval(obj)
    c = []

    for i in obj:
        if i["job"] == "Director" or i["job"] == "Writer":
            c.append(i["job"] + ":" + i["name"])
    return c
```

```python
data["genres"] = data["genres"].apply(extract_1)
```

```python
data["keywords"] = data["keywords"].apply(extract_1)
```

```python
data["cast"] = data["cast"].apply(extract_2)
```

```python
data["spoken_languages"] = data["spoken_languages"].apply(extract_1)
```

```python
data["crew"] = data["crew"].apply(extract_3)
```

```python
print(data["cast"].iloc[0])
```

```python
def collapse(L):
    return [i.replace(" ", "")for i in L]

list_columns = {
    "genres",
    "keywords",
    "cast",
    "crew",
    "spoken_languages"
}

for col in list_columns:
    data[col] = data[col].apply(collapse)
```

```python
data["genres"] = data["genres"].apply(lambda x : " ".join(x))
data["keywords"] = data["keywords"].apply(lambda x : " ".join(x))
data["spoken_languages"] = data["spoken_languages"].apply(lambda x : " ".join(x))
data["cast"] = data["cast"].apply(lambda x : " ".join(x))
data["crew"] = data["crew"].apply(lambda x : " ".join(x))

data["tags"] = (
    data["genres"] + " " + 
    data["keywords"] + " " +
    data["spoken_languages"] + " " +
    data["cast"] + " " +
    data["crew"] + " " +
    data["overview"]
)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

cv = CountVectorizer(max_features = 5000, stop_words = "english")
vectors = cv.fit_transform(data["tags"])
similarity = cosine_similarity(vectors)
print(similarity)
```

```python
def recommend(movies):
    movie_index = data[data["title_x"].str.lower() == movies.lower()].index[0]

    movie_title = data.iloc[movie_index]["title_x"]

    print("Movie:", movie_title)
    print("Movie index:", movie_index)
    
    movie_similarity = list(
        enumerate(similarity[movie_index])
    )
    
    movie_similarity = sorted(
        movie_similarity,
        key = lambda x : x[1],
        reverse = True
    )
    for index,score in movie_similarity[1:6]:
        print(
            data.iloc[index]["title_x"],
            "->",
            round (score, 3)
        )
```
